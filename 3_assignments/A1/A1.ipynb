```python
#Problem 2.3
taxrevModel = sm.OLS(df['ltaxrev'],df[['taxrate','const']])
taxrevResult = taxrevModel.fit()
print(taxrevResult.summary())
```

```python
#Problem 2.5
df['lpop'] = np.log(df['pop'])
taxrevModel2 = sm.OLS(df['ltaxrev'],df[['taxrate','lpop','const']])
taxrevResult2 = taxrevModel2.fit()
print(taxrevResult2.summary())
```

```python
#Problem 3.3
# Regression of taxrate on log(pop)
model = sm.OLS(df['taxrate'],df[['lpop','const']])
result = model.fit()
df['res1'] = result.resid
print(result.summary())
```

```python
# Problem 3.3 continued
# Regresser log(taxrev) på res1
model2 = sm.OLS(df['ltaxrev'],df[['res1','const']])
result2 = model2.fit()
print(result2.summary())
# Giver den samme betahat1 = 0.9711, som da vi lavede regression tidligere 
```

```python
df['taxrevPerBeboer'] = df['taxrev']/df['pop']
testmodel = sm.OLS(df['taxrevPerBeboer'], df[['taxrate','const']])
testResult = testmodel.fit()
print(testResult.summary())
```
